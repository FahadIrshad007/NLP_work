```python
# (a) Extract text from each page of a PDF
import pdfplumber

pdf_path = "sample.pdf"
pages_text = []

with pdfplumber.open(pdf_path) as pdf:
    for page in pdf.pages:
        text = page.extract_text()
        pages_text.append(text if text else "")

print("Total pages extracted:", len(pages_text))
```

```python
# (b) Merge pages and tokenize
import re

full_text = " ".join(pages_text)
cleaned = re.sub(r'[^a-zA-Z\s]', '', full_text)
cleaned = cleaned.lower()
tokens = cleaned.split()

print("Total tokens:", len(tokens))
```

```python
# (c) Word frequency (stopwords removed)
stopwords = ["the","is","in","and","to","of","a","an","for","on","with",
             "as","by","at","from","that","this","it","are","was","be"]

freq = {}
for word in tokens:
    if word not in stopwords:
        freq[word] = freq.get(word, 0) + 1

sorted_freq = sorted(freq.items(), key=lambda x: x[1], reverse=True)

print("Top 10 most frequent words:")
for word, count in sorted_freq[:10]:
    print(word, "-", count)
```

```python
# (d) Keyword search across pages
keyword = input("Enter a keyword to search: ")
found = False

for i, page_text in enumerate(pages_text, start=1):
    sentences = re.split(r'(?<=[.!?]) ', page_text)
    for sentence in sentences:
        if keyword.lower() in sentence.lower():
            print(f"Page {i}: {sentence.strip()}")
            found = True

if not found:
    print("No sentence found containing the keyword.")
```

## Task 2: Structured Page Metadata Extraction and CSV Export

```python
# (a) Build pages_data list of dictionaries
import pdfplumber

pdf_path = "sample.pdf"  # must have at least 5 pages
pages_data = []

with pdfplumber.open(pdf_path) as pdf:
    for i, page in enumerate(pdf.pages, start=1):
        text = page.extract_text()
        text = text if text else ""
        word_count = len(text.split())
        char_count = len(text.replace(" ", ""))
        pages_data.append({
            "page_number": i,
            "raw_text": text,
            "word_count": word_count,
            "char_count": char_count
        })

print(pages_data[:2])
```

```python
# (b) Highest / lowest word count pages and average
non_zero_pages = [p for p in pages_data if p["word_count"] > 0]

max_page = max(pages_data, key=lambda p: p["word_count"])
min_page = min(non_zero_pages, key=lambda p: p["word_count"])
avg_words = round(sum(p["word_count"] for p in pages_data) / len(pages_data), 2)

print("Page with highest word count:", max_page["page_number"])
print(max_page["raw_text"])

print("\nPage with lowest non-zero word count:", min_page["page_number"])
print("Average word count:", avg_words)
```

```python
# (c) Search phrase with custom exception
class PhraseNotFoundError(Exception):
    pass

search_phrase = input("Enter search phrase: ").lower()
match_found = False

try:
    for p in pages_data:
        count = p["raw_text"].lower().count(search_phrase)
        if count > 0:
            print(f"Page {p['page_number']}: {count} occurrence(s)")
            match_found = True

    if not match_found:
        raise PhraseNotFoundError("Phrase not found in any page.")

except PhraseNotFoundError as e:
    print("Error:", e)
```

```python
# (d) Write and re-read pdf_summary.csv
import csv

with open("pdf_summary.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["page_number", "word_count", "char_count"])
    for p in pages_data:
        writer.writerow([p["page_number"], p["word_count"], p["char_count"]])

with open("pdf_summary.csv", "r") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

## Task 3: PDF Text Preprocessing Pipeline for NLP

```python
# (a) Extract text and clean it
import pdfplumber
import re

pdf_path = "sample.pdf"
text = ""

with pdfplumber.open(pdf_path) as pdf:
    for page in pdf.pages:
        page_text = page.extract_text()
        if page_text:
            text += page_text + " "

print("Character count before cleaning:", len(text))

text = re.sub(r'http\S+|www\S+', '', text)      # remove URLs
text = re.sub(r'\d+', '', text)                 # remove digits
text = re.sub(r'[^a-zA-Z\s.]', '', text)         # keep letters, spaces, full stops
text = re.sub(r'\s+', ' ', text).strip()         # collapse spaces

print("Character count after cleaning:", len(text))
```

```python
# (b) Lowercase and split into sentences
text = text.lower()
sentences = re.split(r'[.!?]', text)
sentences = [s.strip() for s in sentences]

print("Sentence count before removing empty:", len(sentences))

sentences = [s for s in sentences if s != ""]

print("Sentence count after removing empty:", len(sentences))
```

```python
# (c) Remove stopwords and filter short sentences
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))
before_count = len(sentences)

filtered_sentences = []
for sentence in sentences:
    words = sentence.split()
    words = [w for w in words if w not in stop_words]
    if len(words) >= 5:
        filtered_sentences.append(" ".join(words))

print("Sentence count before filter:", before_count)
print("Sentence count after filter:", len(filtered_sentences))
print("Sample sentences:", filtered_sentences[:3])
```

```python
# (d) Write to file and analyze
with open("nlp_ready_output.txt", "w") as f:
    for s in filtered_sentences:
        f.write(s + "\n")

with open("nlp_ready_output.txt", "r") as f:
    lines = [line.strip() for line in f.readlines()]

longest = max(lines, key=len)
shortest = min(lines, key=len)

print("Total lines written:", len(lines))
print("Longest sentence:", longest)
print("Shortest sentence:", shortest)
```

## Task 4: DOCX Paragraph Classification and Statistical Analysis

```python
# (a) Read paragraphs and their styles
from docx import Document

doc_path = "sample.docx"
doc = Document(doc_path)

paragraphs_info = []
for para in doc.paragraphs:
    paragraphs_info.append({"text": para.text, "style": para.style.name})

unique_styles = set(p["style"] for p in paragraphs_info)

print("Total paragraph count:", len(paragraphs_info))
print("Unique styles found:", unique_styles)
```

```python
# (b) Classify paragraphs into Headings, Body, Other/Empty
headings = []
body = []
other = []

for p in paragraphs_info:
    style = p["style"]
    if style.startswith("Heading"):
        headings.append(p)
    elif style == "Normal" or "Body" in style:
        body.append(p)
    else:
        other.append(p)

print("Headings:", len(headings))
print("Body:", len(body))
print("Other/Empty:", len(other))
```

```python
# (c) Word/char stats for Body paragraphs with more than 5 words
qualifying = []
for p in body:
    word_count = len(p["text"].split())
    if word_count > 5:
        char_count = len(p["text"].replace(" ", ""))
        qualifying.append({"text": p["text"], "word_count": word_count, "char_count": char_count})

max_para = max(qualifying, key=lambda p: p["word_count"])
min_para = min(qualifying, key=lambda p: p["word_count"])
mean_words = round(sum(p["word_count"] for p in qualifying) / len(qualifying), 2)

print("Highest word count paragraph:", max_para["text"])
print("Lowest word count paragraph (above 5):", min_para["text"])
print("Mean word count:", mean_words)
```

```python
# (d) Nested dictionary of styles, sorted by count
import json

style_dict = {}
for p in paragraphs_info:
    style = p["style"]
    if style not in style_dict:
        style_dict[style] = {"count": 0, "texts": []}
    style_dict[style]["count"] += 1
    if len(style_dict[style]["texts"]) < 2:
        style_dict[style]["texts"].append(p["text"])

sorted_styles = dict(sorted(style_dict.items(), key=lambda x: x[1]["count"], reverse=True))

print(json.dumps(sorted_styles, indent=2))
```

## Task 5: DOCX Table Extraction and Pandas Integration

```python
# (a) Extract all tables from a docx file
from docx import Document

doc_path = "sample_with_table.docx"
doc = Document(doc_path)

all_tables = []
for table in doc.tables:
    table_data = []
    for row in table.rows:
        row_data = [cell.text for cell in row.cells]
        table_data.append(row_data)
    all_tables.append(table_data)

print("Total tables found:", len(all_tables))
for i, t in enumerate(all_tables, start=1):
    print(f"Table {i} dimensions: {len(t)} rows x {len(t[0])} columns")
```

```python
# (b) First table into a pandas DataFrame
import pandas as pd

first_table = all_tables[0]
header = first_table[0]
data_rows = first_table[1:]

df = pd.DataFrame(data_rows, columns=header)
print(df)

empty_counts = df.isnull().sum() + (df == "").sum()
print("\nEmpty or None cells per column:")
print(empty_counts)
```

```python
# (c) Paragraphs before the first table
from docx.oxml.ns import qn

pre_table_paragraphs = []
for child in doc.element.body:
    if child.tag == qn('w:tbl'):
        break
    if child.tag == qn('w:p'):
        text = "".join(node.text for node in child.iter() if node.tag == qn('w:t') and node.text)
        pre_table_paragraphs.append(text)

print("Paragraphs before first table:", len(pre_table_paragraphs))
```

```python
# (d) Write pre-table paragraphs and table data to a text file
with open("docx_extracted.txt", "w") as f:
    f.write("--- PRE-TABLE PARAGRAPHS ---\n")
    for p in pre_table_paragraphs:
        f.write(p + "\n")

    f.write("\n--- TABLE DATA ---\n")
    for row in first_table:
        f.write(" | ".join(row) + "\n")

print("File docx_extracted.txt created successfully.")
```

## Task 6: Multi-Document Comparison, Deduplication, and Frequency Report

```python
# (a) Compare Heading paragraphs from two documents
from docx import Document

doc_a = Document("document_a.docx")
doc_b = Document("document_b.docx")

headings_a = [p.text for p in doc_a.paragraphs if p.style.name.startswith("Heading")]
headings_b = [p.text for p in doc_b.paragraphs if p.style.name.startswith("Heading")]

set_a = set(headings_a)
set_b = set(headings_b)

print("Common headings:", set_a & set_b)
print("Only in Document A:", set_a - set_b)
print("Only in Document B:", set_b - set_a)
```

```python
# (b) Merge Normal paragraphs and remove duplicates
normal_a = [p.text for p in doc_a.paragraphs if p.style.name == "Normal"]
normal_b = [p.text for p in doc_b.paragraphs if p.style.name == "Normal"]

combined = normal_a + normal_b
print("Total paragraphs before deduplication:", len(combined))

seen = set()
deduped = []
for text in combined:
    key = text.strip().lower()
    if key not in seen:
        seen.add(key)
        deduped.append(text)

print("Total paragraphs after deduplication:", len(deduped))
```

```python
# (c) Combined word frequency using Counter
import re
from collections import Counter

stopwords_list = ["the","is","in","and","to","of","a","an","for","on",
                   "with","as","by","at","from","that","this","it","are","was"]

all_words = []
for text in deduped:
    cleaned = re.sub(r'[^a-zA-Z\s]', '', text).lower()
    words = cleaned.split()
    all_words.extend(words)

word_freq = Counter(all_words)
for word in stopwords_list:
    if word in word_freq:
        del word_freq[word]

print("Top 20 most frequent words:")
for word, count in word_freq.most_common(20):
    print(word, "-", count)
```

```python
# (d) Summary statistics
total_headings = len(headings_a) + len(headings_b)
total_unique_body = len(deduped)
vocab_size = len(word_freq)
avg_length = round(len(all_words) / len(deduped), 2)
longest_para = max(deduped, key=lambda t: len(t.split()))

print("Total headings:", total_headings)
print("Total unique body paragraphs:", total_unique_body)
print("Vocabulary size:", vocab_size)
print("Average paragraph length (words):", avg_length)
print("Longest paragraph:", longest_para)
```

## Task 7: Multi-Endpoint API Data Joining and Analysis

```python
# (a) Fetch users and posts
import requests

users_res = requests.get("https://jsonplaceholder.typicode.com/users")
posts_res = requests.get("https://jsonplaceholder.typicode.com/posts")

if users_res.status_code != 200:
    raise ConnectionError(f"Users request failed with status {users_res.status_code}")
if posts_res.status_code != 200:
    raise ConnectionError(f"Posts request failed with status {posts_res.status_code}")

users = users_res.json()
posts = posts_res.json()

print("Users fetched:", len(users))
print("Posts fetched:", len(posts))
```

```python
# (b) Join users and posts
joined = {}
for user in users:
    user_posts = [p["title"] for p in posts if p["userId"] == user["id"]]
    joined[user["name"]] = {
        "email": user["email"],
        "post_count": len(user_posts),
        "titles": user_posts
    }

for name in list(joined.keys())[:3]:
    print(name, "->", joined[name])
```

```python
# (c) User with the most posts
top_user_name = max(joined, key=lambda n: joined[n]["post_count"])
top_user_data = joined[top_user_name]
top_user_full = next(u for u in users if u["name"] == top_user_name)

print("Full name:", top_user_name)
print("Email:", top_user_data["email"])
print("Company:", top_user_full["company"]["name"])
print("Post titles:")
for i, title in enumerate(top_user_data["titles"], start=1):
    print(i, "-", title)
```

```python
# (d) Filter users by company bs or city length
print("Users matching filter conditions:")
for user in users:
    bs = user["company"]["bs"].lower()
    city = user["address"]["city"]
    if "synergies" in bs or len(city) > 8:
        post_count = joined[user["name"]]["post_count"]
        print(user["name"], "-", city, "-", post_count, "posts")
```

## Task 8: To-Do Completion Rate Analysis Across Users

```python
# (a) Fetch and validate todos
import requests

res = requests.get("https://jsonplaceholder.typicode.com/todos")

if res.status_code != 200:
    print("Error: could not fetch data")
    exit()

todos = res.json()

if not isinstance(todos, list):
    print("Error: response is not a list")
    exit()

if len(todos) < 1:
    print("Error: no records found")
    exit()

print("Total todo records:", len(todos))
```

```python
# (b) Group todos by userId
user_stats = {}

for todo in todos:
    uid = todo["userId"]
    if uid not in user_stats:
        user_stats[uid] = {"total": 0, "completed": 0, "incomplete": 0}
    user_stats[uid]["total"] += 1
    if todo["completed"]:
        user_stats[uid]["completed"] += 1
    else:
        user_stats[uid]["incomplete"] += 1

for uid, stats in user_stats.items():
    check = (stats["completed"] + stats["incomplete"]) == stats["total"]
    print(uid, stats, "Check:", check)
```

```python
# (c) Completion rate per user
rates = {}
for uid, stats in user_stats.items():
    rate = round((stats["completed"] / stats["total"]) * 100, 1)
    rates[uid] = rate

highest_uid = max(rates, key=rates.get)
lowest_uid = min(rates, key=rates.get)
avg_rate = round(sum(rates.values()) / len(rates), 1)

print("Highest completion rate:", highest_uid, "-", rates[highest_uid])
print("Lowest completion rate:", lowest_uid, "-", rates[lowest_uid])
print("Average completion rate:", avg_rate)
```

```python
# (d) Formatted ranking table
users_res = requests.get("https://jsonplaceholder.typicode.com/users")
users = users_res.json()
name_lookup = {u["id"]: u["name"] for u in users}

sorted_rates = sorted(rates.items(), key=lambda x: x[1], reverse=True)

print(f"{'Rank':<5}{'Name':<20}{'Total Tasks':<13}{'Completed':<11}{'Completion Rate %':<18}")
for rank, (uid, rate) in enumerate(sorted_rates, start=1):
    name = name_lookup[uid]
    total = user_stats[uid]["total"]
    completed = user_stats[uid]["completed"]
    print(f"{rank:<5}{name:<20}{total:<13}{completed:<11}{rate:<18}")
```

## Task 9: Nested JSON Flattening, DataFrame Operations, and Comment Analysis

```python
# (a) Flatten nested user JSON
import requests

users_res = requests.get("https://jsonplaceholder.typicode.com/users")
users = users_res.json()

def flatten_user(user):
    return {
        "id": user["id"],
        "name": user["name"],
        "username": user["username"],
        "email": user["email"],
        "city": user["address"]["city"],
        "zipcode": user["address"]["zipcode"],
        "lat": user["address"]["geo"]["lat"],
        "lng": user["address"]["geo"]["lng"],
        "company_name": user["company"]["name"],
        "company_catchphrase": user["company"]["catchPhrase"]
    }

flattened_users = [flatten_user(u) for u in users]
print(flattened_users)
```

```python
# (b) DataFrame operations and distance calculation
import pandas as pd
import math

df = pd.DataFrame(flattened_users)
print(df)

df["lat"] = df["lat"].astype(float)
df["lng"] = df["lng"].astype(float)

df["distance_from_origin"] = df.apply(lambda row: math.sqrt(row["lat"]**2 + row["lng"]**2), axis=1)

farthest_user = df.loc[df["distance_from_origin"].idxmax()]
print("\nUser with greatest distance from origin:")
print(farthest_user)
```

```python
# (c) Hemisphere classification
northern = df[df["lat"] > 0]
southern = df[df["lat"] < 0]

print("Northern Hemisphere users:", list(northern["name"]))
print("Southern Hemisphere users:", list(southern["name"]))

north_avg = northern["distance_from_origin"].mean()
south_avg = southern["distance_from_origin"].mean()

if north_avg > south_avg:
    print("Northern Hemisphere has higher average distance.")
else:
    print("Southern Hemisphere has higher average distance.")
```

```python
# (d) Comment counts per post
posts_res = requests.get("https://jsonplaceholder.typicode.com/posts")
comments_res = requests.get("https://jsonplaceholder.typicode.com/comments")

posts = posts_res.json()
comments = comments_res.json()

post_comment_counts = []
for post in posts:
    count = sum(1 for c in comments if c["postId"] == post["id"])
    post_comment_counts.append((post["id"], post["title"][:40], count))

post_comment_counts.sort(key=lambda x: x[2], reverse=True)

print("Top 5 most-commented posts:")
for p in post_comment_counts[:5]:
    print(p)

print("\nBottom 5 least-commented posts:")
for p in post_comment_counts[-5:]:
    print(p)
```
